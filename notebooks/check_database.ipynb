```python
db_con = conn
# 1. Check what tables exist in the database
print("=== DATABASE TABLES ===")
tables = db_con.execute("SHOW TABLES").fetchall()
print(f"Tables in database: {tables}")

```

```python
# 2. Check the structure of the map_layers table
print("\n=== MAP_LAYERS TABLE STRUCTURE ===")
schema = db_con.execute("DESCRIBE osm_amenities").fetchall()
for row in schema:
    print(f"Column: {row[0]}, Type: {row[1]}")
```

```python
# 3. Count total number of stored layers
print("\n=== LAYER COUNT ===")
count = db_con.execute("SELECT COUNT(*) FROM osm_amenities").fetchone()
print(f"Total layers stored: {count[0]}")
```

```python
# 4. List all stored layers with basic info
print("\n=== STORED LAYERS ===")
layers_info = db_con.execute("SELECT layer_name, source FROM osm_amenities").fetchall()
for layer_name, source in layers_info:
    print(f"Layer: {layer_name}")
    print(f"  Source: {source}")
```

```python
# 5. Get detailed info about each layer using your Data_Collection methods
print("\n=== DETAILED LAYER INFORMATION ===")
from rampa.duckdb.operations import get_all_layers

all_layers = get_all_layers(db_con)
for layer_info in all_layers:
    layer_data = layer_info['layer_data']
    print(f"\nLayer: {layer_info['layer_name']}")
    print(f"  Record Count: {layer_data.get('record_count', 'Unknown')}")
    print(f"  Columns: {layer_data.get('columns', 'Unknown')}")
    print(f"  CRS: {layer_data.get('crs', 'Unknown')}")
    print(f"  Layer Type: {layer_data.get('layer_type', 'Unknown')}")
    print(f"  Bounds: {layer_data.get('bounds', 'Unknown')}")
```

```python
# Check specifically for demographic indicator layers
print("\n=== DEMOGRAPHIC INDICATOR LAYERS ===")
from rampa.duckdb.operations import get_all_layers

# Define the layers we're looking for
target_layers = [
    'DENSIDAD_POBLACION', 'EDAD_PROMEDIO', 'PROPORCION_JUVENTUD',
    'PROPORCION_ENVEJECIMIENTO', 'PROPORCION_SOBREENVEJECIMIENTO',
    'INDICE_ENVEJECIMIENTO', 'INDICE_JUVENTUD', 'INDICE_DEPENDENCIA',
    'INDICE_ESTRUCTURA_POBLACION_ACTIVA', 'INDICE_REEMPLAZO_POBLACION_ACTIVA',
    'RAZON_PROGRESIVIDAD_DEMOGRAFICA'
]

all_layers = get_all_layers(db_con)
found_layers = []

for layer_info in all_layers:
    layer_name = layer_info['layer_name']
    if layer_name in target_layers:
        found_layers.append(layer_name)
        layer_data = layer_info['layer_data']
        print(f"\nLayer: {layer_name}")
        print(f"  Record Count: {layer_data.get('record_count', 'Unknown')}")
        print(f"  Columns: {layer_data.get('columns', 'Unknown')}")
        print(f"  CRS: {layer_data.get('crs', 'Unknown')}")
        print(f"  Layer Type: {layer_data.get('layer_type', 'Unknown')}")

print(f"\nFound {len(found_layers)} out of {len(target_layers)} target layers:")
print(f"Found: {found_layers}")

missing_layers = [layer for layer in target_layers if layer not in found_layers]
print(f"Missing: {missing_layers}")

# Also check for similar layer names
print(f"\n=== SIMILAR LAYER NAMES ===")
all_layer_names = [layer_info['layer_name'] for layer_info in all_layers]
for target in target_layers:
    similar = [name for name in all_layer_names if target.lower() in name.lower() or name.lower() in target.lower()]
    if similar:
        print(f"{target}: Found similar -> {similar}")
```
