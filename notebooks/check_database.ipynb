```python
import duckdb
conn = duckdb.connect('/Users/admin/Desktop/personal_projects/Madrid Open Data/madrid-open-data/rampa/duckdb/databases/madrid_layers.db')
```

```python
conn.execute("SHOW TABLES").fetchall()
```

```python
db_con = conn
# 1. Check what tables exist in the database
print("=== DATABASE TABLES ===")
tables = db_con.execute("SHOW TABLES").fetchall()
print(f"Tables in database: {tables}")

```

```python
# 2. Check the structure of the map_layers table
print("\n=== MAP_LAYERS TABLE STRUCTURE ===")
schema = db_con.execute("DESCRIBE map_layers").fetchall()
for row in schema:
    print(f"Column: {row[0]}, Type: {row[1]}")
```

```python
# 3. Count total number of stored layers
print("\n=== LAYER COUNT ===")
count = db_con.execute("SELECT COUNT(*) FROM map_layers").fetchone()
print(f"Total layers stored: {count[0]}")
```

```python
# 4. List all stored layers with basic info
print("\n=== STORED LAYERS ===")
layers_info = db_con.execute("SELECT layer_name, source FROM map_layers").fetchall()
for layer_name, source in layers_info:
    print(f"Layer: {layer_name}")
    print(f"  Source: {source}")
```

```python
# 5. Get detailed info about each layer using your Data_Collection methods
print("\n=== DETAILED LAYER INFORMATION ===")
from rampa.duckdb.operations import get_all_layers

all_layers = get_all_layers(db_con)
for layer_info in all_layers:
    layer_data = layer_info['layer_data']
    print(f"\nLayer: {layer_info['layer_name']}")
    print(f"  Record Count: {layer_data.get('record_count', 'Unknown')}")
    print(f"  Columns: {layer_data.get('columns', 'Unknown')}")
    print(f"  CRS: {layer_data.get('crs', 'Unknown')}")
    print(f"  Layer Type: {layer_data.get('layer_type', 'Unknown')}")
    print(f"  Bounds: {layer_data.get('bounds', 'Unknown')}")
```

```python
# 6. Check database file size (if using file-based DB)
import os
db_path = "/Users/admin/Desktop/personal_projects/Madrid Open Data/madrid-open-data/rampa/duckdb/databases/madrid_layers.db"
if os.path.exists(db_path):
    size_mb = os.path.getsize(db_path) / (1024 * 1024)
    print(f"\n=== DATABASE FILE INFO ===")
    print(f"Database file size: {size_mb:.2f} MB")
```

```python
# 7. Get memory usage info (for in-memory operations)
print("\n=== DATABASE MEMORY INFO ===")
try:
    memory_info = db_con.execute("PRAGMA database_list").fetchall()
    for db_info in memory_info:
        print(f"Database: {db_info}")
except:
    print("Memory info not available")
```
